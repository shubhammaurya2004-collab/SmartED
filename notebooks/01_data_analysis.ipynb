```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the hospital dataset
df = pd.read_csv("../data/hospital_waiting.csv")

# Display the first 5 rows
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(df["waiting_time"], bins=30, kde=True)

plt.title("Distribution of Hospital Waiting Time")
plt.xlabel("Waiting Time (minutes)")
plt.ylabel("Number of Patients")

plt.show()
```

Patients waiting and  Waiting time

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="patients_waiting",
    y="waiting_time",
    alpha=0.4
)

plt.title("Patients Waiting vs Hospital Waiting Time")
plt.xlabel("Number of Patients Waiting")
plt.ylabel("Waiting Time (minutes)")

plt.show()
```

Doctors available vs waiting time


```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="doctors_available",
    y="waiting_time"
)

plt.title("Doctors Available vs Hospital Waiting Time")
plt.xlabel("Number of Doctors Available")
plt.ylabel("Waiting Time (minutes)")

plt.show()
```

Hospital occupancy vs waiting time

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="hospital_occupancy",
    y="waiting_time",
    alpha=0.4
)

plt.title("Hospital Occupancy vs Waiting Time")
plt.xlabel("Hospital Occupancy (%)")
plt.ylabel("Waiting Time (minutes)")

plt.show()
```

```python
plt.figure(figsize=(12, 8))

correlation = df.select_dtypes(include="number").corr()

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Hospital Variables")

plt.show()
```
